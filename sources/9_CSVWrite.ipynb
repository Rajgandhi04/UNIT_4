```python
import csv
```

```python
makes = ['Toyota', 'Honda', 'Ford', 'Chevrolet', 'Nissan']
models = ['Camry', 'Civic', 'F-150', 'Silverado', 'Altima']
max_speeds = [120, 130, 110, 115, 125]
powers = [200, 180, 250, 300, 220]
```

```python
cars_list = list(zip(makes, models, max_speeds, powers))
```

```python
cars_list
```

```python
with open('../datasets/cars_data.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['Make', 'Model', 'Max Speed', 'Power'])
    writer.writerows(cars_list)
```
